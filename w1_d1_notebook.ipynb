```python

from openai import OpenAI

# Store conversation history
messages = []

# Create the Ollama client once
ollama = OpenAI(base_url=os.getenv("ollama_api_key"),api_key="ollama")

def chat(text):
    messages.append({
        "role": "user",
        "content": text
    })

    response = ollama.chat.completions.create(model="llama3.2:3b",messages=messages,temperature=1)
    res = response.choices[0].message.content

    messages.append({
        "role": "assistant",
        "content": res
    })

    return res



print(chat("My name is Raja."))
print(chat("What is my name?"))

```

```python
user_prompt="""see,now i am going to provide an website url to you ,once go through that website and give me a clear 
insights of it..and what does the website consist of and how it is useful to students.."""


system_prompt="""you are an clear explanier of each topic,
make sure you markdown every key feature in it with deatailed explination,clealry with examples after each key feature."""

def messages_(url):
    message= [
        {"role":"user","content":user_prompt+url},
        {"role":"system","content":system_prompt}
    ]
    return message
```

```python
def web_summarizer(url):
   res=ollama.chat.completions.create(model="llama3.2:3b",messages=messages_(url))
   return res.choices[0].message.content
```

```python
from urllib.parse import urlparse, parse_qs

def extract_video_id(url):
    parsed = urlparse(url.strip())
    hostname = (parsed.hostname or "").lower()

    if hostname in {
        "youtube.com",
        "www.youtube.com",
        "m.youtube.com"
    }:
        video_id = parse_qs(parsed.query).get("v", [None])[0]

    elif hostname == "youtu.be":
        video_id = parsed.path.strip("/").split("/")[0]

    elif hostname in {
        "youtube-nocookie.com",
        "www.youtube-nocookie.com"
    }:
        parts = parsed.path.strip("/").split("/")
        video_id = parts[-1] if parts else None

    else:
        raise ValueError("Please provide a supported YouTube URL.")

    if not video_id or len(video_id) != 11:
        raise ValueError("Invalid or unsupported YouTube video URL.")

    return video_id
```

```python
from youtube_transcript_api import YouTubeTranscriptApi
api = YouTubeTranscriptApi()


def get_transcript(video_id):#the function is used to retive the cpations for a given YT video.

    try:
        transcript_list = api.list(video_id)#fecthes the avaiable captions and all metadata for the given YT video.

        available_languages = [transcript.language_code for transcript in transcript_list]

        print("Available languages:", available_languages)

        transcript = api.fetch(video_id,languages=["en", "te", "hi"])#retrives the actual transcript available in that engilsh.

        transcript_text = " ".join(snippet.text for snippet in transcript)

        if not transcript_text.strip():
            raise ValueError("The transcript is empty.")

        return transcript_text

    except Exception as error:
        print("Transcript retrieval failed:")
        print(type(error).__name__, error)
        return None
```

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

ollama = OpenAI(base_url=os.getenv("ollama_api_key"),api_key="ollama")
gemini_key=OpenAI(base_url="https://generativelanguage.googleapis.com/v1beta/openai/",api_key=os.getenv("gemini_api_key"))
```

```python
system_prompt="""
You are an expert YouTube learning assistant. Analyze the provided video transcript and create a structured, beginner-friendly learning resource.

## 1. Video Overview

Briefly explain the video's topic, objectives, and intended audience.

## 2. Detailed Summary

Summarize the transcript in logical sections, following the flow of ideas. Explain important concepts clearly and use timestamps when available.

## 3. Key Takeaways

List the most important lessons, insights, and actionable points.

## 4. Important Concepts Explained Simply

Identify the major concepts and explain each using simple language, examples, analogies, and practical applications.

## 5. Practical Examples

Provide relevant real-world examples, use cases, code snippets, or exercises where appropriate. Clearly distinguish illustrative examples from transcript content.

## 6. Five Quiz Questions with Answers

Generate five questions testing conceptual understanding and practical application. Include the correct answer and a brief explanation for each.

## 7. Learning Roadmap and Related YouTube Videos

Create an interview-oriented learning path with three levels:

* **Level 1 — Fundamentals:** 3–5 beginner videos covering prerequisites and core concepts.
* **Level 2 — Intermediate:** 3–5 videos covering deeper concepts, implementations, and practical projects.
* **Level 3 — Advanced:** 3–5 videos covering advanced techniques, real-world projects, optimization, architecture, and interview preparation where relevant.

For every recommended video, provide:

* Watch order
* Verified video title and channel name
* Difficulty level
* Key concepts covered
* Relevance to the original transcript
* Expected learning outcome
* Clickable YouTube link in this exact format: `[Video Title](https://www.youtube.com/watch?v=VIDEO_ID)`

Prioritize reputable educational channels and relevant, up-to-date resources.

**Recommendation rules:**

* Never invent video titles, channel names, video IDs, URLs, or video content.
* Use verified search results or trusted video metadata when available.
* If video verification is unavailable, explicitly state the limitation rather than presenting recommendations as verified.
* Avoid duplicate or unrelated videos.
* If fewer relevant videos can be verified, recommend fewer.

Finish with a numbered watch sequence showing how each video builds toward practical skills and interview readiness.

## 8. Revision Checklist

Summarize the essential concepts, practical skills, exercises, and next learning steps.

## Global Rules

* Base the original video's analysis strictly on its transcript.
* Distinguish transcript information from additional explanations and illustrative examples.
* Use clear English, concise explanations, and well-structured Markdown.
* Adapt the roadmap to the actual topic; avoid irrelevant sections.
* Ensure YouTube links are formatted as clickable Markdown links.
* Never claim to have searched for or verified resources unless you actually have.


"""
messages=[{"role":"user","content":transcript_text},{"role":"system","content":system_prompt}]
```

```python
def summarize_video_ollama(text):

    response = ollama.chat.completions.create(model="llama3.2:3b",messages=messages)
    return response.choices[0].message.content


def summarize_video_gemini(text):
    res=gemini_key.chat.completions.create(model="gemini-3.5-flash",messages=messages,temperature=0.5)
    return res.choices[0].message.content

```
