```python
from urllib.parse import urlparse, parse_qs

def extract_video_id(url):
    parsed = urlparse(url.strip())
    hostname = (parsed.hostname or "").lower()

    if hostname in {
        "youtube.com",
        "www.youtube.com",
        "m.youtube.com"
    }:
        video_id = parse_qs(parsed.query).get("v", [None])[0]

    elif hostname == "youtu.be":
        video_id = parsed.path.strip("/").split("/")[0]

    elif hostname in {
        "youtube-nocookie.com",
        "www.youtube-nocookie.com"
    }:
        parts = parsed.path.strip("/").split("/")
        video_id = parts[-1] if parts else None

    else:
        raise ValueError("Please provide a supported YouTube URL.")

    if not video_id or len(video_id) != 11:
        raise ValueError("Invalid or unsupported YouTube video URL.")

    return video_id
```

```python
from youtube_transcript_api import YouTubeTranscriptApi
api = YouTubeTranscriptApi()


def get_transcript(video_id):#the function is used to retive the cpations for a given YT video.

    try:
        transcript_list = api.list(video_id)#fecthes the avaiable captions and all metadata for the given YT video.

        available_languages = [transcript.language_code for transcript in transcript_list]

        print("Available languages:", available_languages)

        transcript = api.fetch(video_id,languages=["en", "te", "hi"])#retrives the actual transcript available in that engilsh.

        transcript_text = " ".join(snippet.text for snippet in transcript)

        if not transcript_text.strip():
            raise ValueError("The transcript is empty.")

        return transcript_text

    except Exception as error:
        print("Transcript retrieval failed:")
        print(type(error).__name__, error)
        return None
```

```python
import gradio as gr

def analyze_video(video_url):
    if not video_url or not video_url.strip():
        raise gr.Error("Please enter a YouTube URL.")

    try:
        # Reuse your existing functions here.
        video_id = extract_video_id(video_url.strip())

        transcript = get_transcript(video_id)

        if not transcript:
            raise gr.Error("No transcript was found.")

        # If your transcript generator returns a list or dictionary,
        # adapt this line to obtain the actual transcript text.
        if isinstance(transcript, list):
            transcript = " ".join(
                item["text"] for item in transcript
            )

        learning_guide = chain.invoke({
            "transcript": transcript
        })

        return learning_guide, transcript

    except gr.Error:
        raise

    except Exception as exc:
        raise gr.Error(
            f"Unable to process the video: {exc}"
        ) from exc


with gr.Blocks(
    theme=gr.themes.Soft(),
    title="YouTube Learning Assistant"
) as demo:

    gr.Markdown(
        """
        # YouTube Learning Assistant

        Transform YouTube videos into structured learning guides.

        Paste a video URL to explore the summary, key concepts,
        practical examples, and knowledge quiz.
        """
    )

    with gr.Row():
        url_input = gr.Textbox(
            label="YouTube Video URL",
            placeholder="Paste your YouTube URL here",
            scale=4
        )

        analyze_button = gr.Button(
            "Analyze Video",
            variant="primary"
        )

    with gr.Tabs():
        with gr.Tab("Learning Guide"):
            summary_output = gr.Markdown()

        with gr.Tab("Transcript"):
            transcript_output = gr.Textbox(
                label="Extracted Transcript",
                lines=15,
                interactive=False
            )

    analyze_button.click(
        fn=analyze_video,
        inputs=url_input,
        outputs=[summary_output, transcript_output]
    )

    url_input.submit(
        fn=analyze_video,
        inputs=url_input,
        outputs=[summary_output, transcript_output]
    )
```

```python
demo.queue().launch()
```
